```python
class Currency_Converter:
    def start(self):
        print("Welcome to Currency Converter:")
        print("Please press the options for the conversion:")
        print("1: Convert your money from PKR to USD $")
        print("2: Convert your money from USD $ to PKR")
        print("3: Convert your money from PKR to Euros")
        print("4: Convert your money from Euros to PKR")
        print("5: Convert your money from PKR to Pounds")
        print("6: Convert your money from Pounds to PKR")
        print("7: Convert your money from PKR to Riyal")
        print("8: Convert your money from Riyal to PKR")
        
        choice = int(input("Enter your choice (1/2/3/4/5/6/7/8): "))
        amount = float(input("Enter the amount: "))
        
        currency_names = {
            "PKR": "Pakistani Rupees",
            "USD $": "United States Dollars",
            "Euros": "Euros",
            "Pounds": "British Pounds",
            "Riyal": "Saudi Riyals"
        }

        if choice == 1:
            converted_amount = amount * 0.0035  # PKR to USD $
            source_currency, target_currency = "PKR", "USD $"
        elif choice == 2:
            converted_amount = amount * 285.71  # USD $ to PKR
            source_currency, target_currency = "USD $", "PKR"
        elif choice == 3:
            converted_amount = amount * 0.0032  # PKR to Euros
            source_currency, target_currency = "PKR", "Euros"
        elif choice == 4:
            converted_amount = amount * 316.64  # Euros to PKR
            source_currency, target_currency = "Euros", "PKR"
        elif choice == 5:
            converted_amount = amount * 0.0027  # PKR to Pounds
            source_currency, target_currency = "PKR", "Pounds"
        elif choice == 6:
            converted_amount = amount * 367.50  # Pounds to PKR
            source_currency, target_currency = "Pounds", "PKR"
        elif choice == 7:
            converted_amount = amount * 0.013  # PKR to Riyal
            source_currency, target_currency = "PKR", "Riyal"
        elif choice == 8:
            converted_amount = amount * 75.58  # Riyal to PKR
            source_currency, target_currency = "Riyal", "PKR"
        else:
            print("Invalid choice.")
            return

        print(f"{amount} {currency_names[source_currency]} is equal to {converted_amount:.2f} {currency_names[target_currency]}.")

converter = Currency_Converter()
converter.start()
```
