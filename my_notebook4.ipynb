```python
#Matrix Project
rows1 = int(input("Enter the number of rows for matrix 1: "))
cols1 = int(input("Enter the number of columns for matrix 1: "))

rows2 = int(input("Enter the number of rows for matrix 2: "))
cols2 = int(input("Enter the number of columns for matrix 2: "))

print("Enter matrix 1:")
matrix1 = []
for i in range(rows1):
    row = []
    for j in range(cols1):
        value = int(input("Enter the value for row {} and column {}: ".format(i+1, j+1)))
        row.append(value)
    matrix1.append(row)

print("Enter matrix 2:")
matrix2 = []
for i in range(rows2):
    row = []
    for j in range(cols2):
        value = int(input("Enter the value for row {} and column {}: ".format(i+1, j+1)))
        row.append(value)
    matrix2.append(row)

print("Matrix Calculator")
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Exit")

choice = int(input("Enter your choice (1-3): "))

if choice == 1:
    if rows1 != rows2 or cols1 != cols2:
        print("The matrices are not compatible for addition.")
    else:
        result = []
        for i in range(rows1):
            row = []
            for j in range(cols1):
                value = matrix1[i][j] + matrix2[i][j]
                row.append(value)
            result.append(row)
        print("The sum of the matrices is:")
elif choice == 2:
    if rows1 != rows2 or cols1 != cols2:
        print("The matrices are not compatible for subtraction.")
    else:
        result = []
        for i in range(rows1):
            row = []
            for j in range(cols1):
                value = matrix1[i][j] - matrix2[i][j]
                row.append(value)
            result.append(row)
        print("The difference of the matrices is:")
elif choice == 3:
    if cols1 != rows2:
        print("The matrices are not compatible for multiplication.")
    else:
        result = []
        for i in range(rows1):
            row = []
            for j in range(cols2):
                value = 0
                for k in range(cols1):
                    value += matrix1[i][k] * matrix2[k][j]
                row.append(value)
            result.append(row)
        print("The product of the matrices is:")
else:
    print("Invalid choice.")

if choice == 1 or choice == 2 or (choice == 3 and cols1 == rows2):
    for row in result:
        for value in row:
            print(value, end=" ")
        print()
```
