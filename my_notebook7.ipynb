```python
# Aggregiation:
class Publication:
    def get_data(self):
        self.title = input("Enter a title: ")
        self.price = float(input("Enter a price: "))
    def put_data(self):
        print('Title:',self.title)
        print('Price:',self.price)
```

```python
class Sales:
    def get_data(self):
        month1 = float(input("Enter sales amount for month 1: "))
        month2 = float(input("Enter sales amount for month 2: "))
        month3 = float(input("Enter sales amount for month 3: "))
        self.sales = [month1,month2,month3]
        
    def put_data(self):
        print('Sales of 3 Months: ',self.sales)
```

```python
class Book(Publication ):
    def get_data(self,sale):
        sale.get_data()
        Publication.get_data(self)
        self.page_count = int(input(("Enter page count: ")))
        
    def put_data(self,sale):
        sale.put_data()
        Publication.put_data(self)
        print("Page Count:",self.page_count)
```

```python
# Aggregiation:
sale1 = Sales() # sale class
book1 = Book() # book class
book1.get_data(sale1)
```

```python
book1.put_data(sale1)
```
