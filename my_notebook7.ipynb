Define a class called BasicSalary having one instance variable basic to store basic salary of an employee. Add
appropriate __init__ method and another method annualBasicSalary which calculates and returns the sum
of basic salaries for twelve months. Now define another class called Employee which instantiates an object of
BasicSalary type in its __init__ method. It also instantiate an instance variable called annualBonus. Add a

basic and annual bonus.Define a class called BasicSalary having one instance variable basic to store basic salary of an employee. Add
appropriate __init__ method and another method annualBasicSalary which calculates and returns the sum
of basic salaries for twelve months. Now define another class called Employee which instantiates an object of
BasicSalary type in its __init__ method. It also instantiate an instance variable called annualBonus. Add a
method annualNetSalary to Employee which returns the annual net salary of an employee including the annual
basic and annual bonus.

```python
class BasicSalary:
    def __init__(self,b):
        self.basic = b
        
    def annual_basic_salary(self):
        salary = 0
        for i in range(12):
            user = float(input('Enter a salary: '))
            salary = salary + user
            
            self.basic_salary = salary
            return self.basic_salary
```

```python
class Employee:
    def __init__(self,ab):
        self.annual_bonus = ab
        self.basic_salary = BasicSalary(1000) # Object
        
    def annual_net_salary(self):
      annual_salary = self.basic_salary.annual_basic_salary()
      annual_net_salary = (annual_salary / 12) + self.annual_bonus + self.basic_salary.basic
      print(annual_net_salary)
        
```

```python
e1 = Employee(3000)
e1.annual_net_salary()
```

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    
    def __str__(self):  
       return str(('X-Coordinate:' , self.x , 'Y-Coordinate: ',self.y ))
    
    def setter(self,x,y):
        self.x = x
        self.y =y
```

```python
class Distance:
    def __init__(self):
        self.p1 = Point(5,8)
        self.p2 = Point(2,6)
        
    def setter(self,p1,p2):
        self.p1 = p1
        self.p2 = p2
        
    def findDistance(self):
        self.dis = ((self.p1.x - self.p2.x)**2 + (self.p1.y - self.p2.y)**2)**0.5
        print('Distance: ', self.dis)
        
    def __sub__(self,d2):
        dist = self.dis - d2.dis
        return dist
    def __add__(self,d2):
        dist = self.dis + d2.dis
        return dist
    def __mul__(self,d2):
        dist = self.dis * d2.dis
        return dist
    def __truediv__(self,d2):
        dist = self.dis / d2.dis
        return dist
```

```python
d1 = Distance()
d1.findDistance()
```

```python
class Polygon:
    def __init__(self):
        self.lop = []
    
    def addPoint(self,lop):
        self.lop = lop
        
    
    @staticmethod
    def distance(x1,x2):
        dist = ((x1.x - x2.x)**2 + (x1.y - x2.y)**2)**0.5
        return dist
    
    def perimeter(self):
        value_dis = 0
        for i in range(len(self.lop)):
            if i != (len(self.lop)-1):
                dis_meth = Polygon.distance(self.lop[i],self.lop[i+1])
                value_dis = dis_meth + value_dis
            else:
                dis_meth = Polygon.distance(self.lop[i],self.lop[0])
                value_dis = dis_meth + value_dis
        return value_dis
                
                
                
```

```python
square = Polygon()
p1 = Point(9,8)
p2 = Point(7,4)
p3 = Point(6,2)
p4 = Point(5,1)
square.addPoint([p1, p2, p3, p4])
square.perimeter()
```
