1. Create a class Fraction having two private attributes numerator and denominator and the
following methods:
 Constructor to initialize the attributes to 1 by default.
 Setters for the two attributes.
 getFraction( ): returns numerator and denominator as a tuple.
 convertDecimal( ): returns decimal equivalent of the fraction.
Print the list of attributes using the __dict__ attribute.


```python
class Fraction:
    def __init__(self, numerator=1, denominator=1):
        self.__numerator = numerator
        self.__denominator = denominator

    def setnumerator(self, numerator):
        self.__numerator = numerator

    def setdenominator(self, denominator):
        self.__denominator = denominator

    def getFraction(self):
        return self.__numerator, self.__denominator

    def convertDecimal(self):
        if self.__denominator !=0:
            return self.__numerator / self.__denominator
        else:
            return "Divison cannot be possible as denominator is zero."
    
    
    
```

```python
fraction1= Fraction(3,7)
```

```python
print(fraction1.getFraction())
```

```python
fraction1.convertDecimal()
```
