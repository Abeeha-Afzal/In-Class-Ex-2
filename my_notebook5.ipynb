```python
# Inheritance:
# 1) All methods and attributes of present class will transfer into child classes.
```

```python
# Parent
class Shapes:
    def __init__(self,color,sides):
        self.color = color
        self.sides = sides
    def getter(self):
        print('Color: ', self.color)
        print('Sides: ', self.sides)
    
```

```python
# Child
class Square(Shapes):
    def __init__(self,color,length,height):
        self.length = length
        self.height = height
        super().__init__(color,4)
```

```python
shape = Shapes('Blue',4)
shape.getter()
```

```python
sq2 = Square('Orange',3,7)
sq2.getter()
```

```python
square = Square('Pink',4)
square.getter()
```

```python
class Hexagon(Shapes):
    pass
```

```python
hexagon = Hexagon('Yellow',6)
hexagon.getter()
```

```python
class Rectangle(Shapes):
    pass
```

```python
rectangle = Rectangle('Red',4)
rectangle.getter()
```
