```python
# Inheritance:
# 1) All methods and attributes of present class will transfer into child classes.
```

```python
# Parent
class Shapes:
    def __init__(self,color,sides):
        self.color = color
        self.sides = sides
    def getter(self):
        print('Color: ', self.color)
        print('Sides: ', self.sides)
    
```

```python
# Child
class Square(Shapes):
    def __init__(self,color,length,height):
        self.length = length
        self.height = height
        super().__init__(color,4)
    def getter(self): #method overriding
        print("Color: ",self.color)
        print("Sides:",self.sides)
        print("Length: ",self.length)
        print("Height: ",self.height)
    def area(self):
        print("Area of square:",self.length *2)
```

```python
class Rectangle(Shapes):
    def __init__(self,color,length,height):
        self.length = length
        self.height = height
        super().__init__(color,4)
    def getter(self): #method overriding
        print("Color: ",self.color)
        print("Sides:",self.sides)
        print("Length: ",self.length)
        print("Height: ",self.height)
    def area(self):
        print("Area of rectangle:",self.length * self.height)
    
```

```python
class Hexagon(Shapes):
    def __init__(self,color,length,height):
        self.length = length
        self.height = height
        super().__init__(color,6)
    def getter(self): #method overriding
        print("Color: ",self.color)
        print("Sides:",self.sides)
        print("Length: ",self.length)
        print("Height: ",self.height)
    def area(self):
        print("Area of Hexagon:", (3 * (3 ** 0.5) / 2) * self.sides** 2)
        
    
```

```python
hexagon = Hexagon('Yellow',2,7)
hexagon.getter()
```

```python
hexagon.area()
```

```python
rectangle = Rectangle('Red',5,7)
rectangle.getter()
```

```python
rectangle.area()
```

```python
# Parent class
class Animal:
    def __init__(self,no_of_legs,tail):
        self.no_of_legs= no_of_legs
        self.tail = tail
        
    def sound(self,sound):
        self.sound = sound
        #print("Sound of animal:",self.sound)
```

```python
# Child class
class Lion(Animal):
    def __init__(self,color,no_of_legs,tail,age,sound):
        self.color = color
        self.age = age
        self.sound = sound
        super().__init__(no_of_legs,tail)
    
    def getter(self):
        print('Legs:', self.no_of_legs)
        print('sound:',self.sound)
        print('color:',self.color)
        print('tail:',self.tail)
```

```python
Lion1 = Lion('Brown',4, True,2,'roar')
Lion1.getter()
```

```python
class Elephant(Animal):
    def __init__(self,color,no_of_legs,tail,age,trunk_length):
        self.color = color
        self.age = age
        self.trunk_length = trunk_length
        
        super().__init__(no_of_legs,tail)
    
    def getter(self):
        print('Legs:', self.no_of_legs)
        print('sound:',self.sound)
        print('color:',self.color)
        print('tail:',self.tail)
        print('Trunk Length:',self.trunk_length)
        
```

```python
Elephant1 = Elephant('Grey',4,True,3,2.5)
Elephant1.getter()
```
