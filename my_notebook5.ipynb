```python
# Inheritance:
# 1) All methods and attributes of present class will transfer into child classes.
```

```python
# Parent
class Shapes:
    def __init__(self,color,sides):
        self.color = color
        self.sides = sides
    def getter(self):
        print('Color: ', self.color)
        print('Sides: ', self.sides)
    
```

```python
# Child
class Square(Shapes):
    pass
```

```python
shape = Shapes('Blue',4)
shape.getter()
```

```python
square = Square('Pink',4)
square.getter()
```
